```python
def movegen(node):
  city_edges = edge_dict()
  return city_edges[node]

def checkOpen(node, open):
  for i in open:
    if node == i[0]:
      return i
  return 0

def checkClosed(node, closed):
  for i in closed:
    if node == i[0]:
      return True
  return False

def editOpen(newtup, open):
  for i in open:
    if newtup[0] == i[0]:
      print('Edit',newtup, i)
      open.remove(i)
      open.append(newtup)
      return open

def reconstructPath(goal, parent, closed):
  path = [goal]
  currentNode = parent
  for i in closed:
    if i[0] == currentNode:
      path = [i[0]] + path
      currentNode = i[1]
  return path

def kernel_boundary(closed,open, goal='G'):
  kernel = []
  boundary = []
  for node in closed:
    flag = True
    if node[0] != goal:
      for nbr in movegen(node[0]):
        if not checkClosed(nbr[0],closed):
          flag = False
          break
      if flag:
        kernel.append(node[0])
      else:
        boundary.append(node[0])
  print ('Kernel', kernel)
  print ('Boundary', boundary)


def Astar(start, goal, w=1, scale=1):
  k = 0
  open, closed = [], []
  g=0
  h=dist_est(start,goal,w,scale)
  f= g+h
  open.append((start, 'NIL', g,h,f))
  while open != []:
    k += 1
    open.sort(key=lambda x:(x[4],x[0]))
    nodeExplore = open.pop(0)
    node = nodeExplore[0]
    closed = [nodeExplore] + closed
    if node == goal:
      print(f'Exploring Node #: {k}, Node: {node[0]}')
      print('Open', open)
      print('Closed', closed)
      kernel_boundary(closed,open)
      return reconstructPath(node,nodeExplore[1], closed), nodeExplore
    for nbr in movegen(node):
      g = nodeExplore[2] + nbr[1]
      h = dist_est(nbr[0],goal,w,scale)
      f = g + h
      openNbr = checkOpen(nbr[0],open)
      if  openNbr != 0:
        if g < openNbr[2]:
          #update the node in open
          open = editOpen((nbr[0],node,g,h,f),open)
      elif checkClosed(nbr[0],closed):
        pass
      else:
        open.append((nbr[0],node,g,h,f))
    if (k == k):
      print(f'Exploring Node #: {k}, Node: {node[0]}')
      print('Open', open)
      print('Closed', closed)
      kernel_boundary(closed,open)
      print(f'\n')
  return "No path found"

print(Astar('S','G',1,10))

```

##Beam Search


```python
def beamSearch(start, goal, w, scale):
  open = [(start,dist_est(start,goal,1,scale))]
  N = start
  bestEver = N
  flag = True
  while flag:
    print(open, bestEver)
    if goal in open:
      return goal
    nbrList=[]
    for i in open:
      nbrList += movegen(i[0][0])
    nbrs = [(i[0], dist_est(i[0],goal,1,scale)) for i in nbrList]
    nbrs.sort(key = lambda x:(x[1],x[0]))
    open = nbrs[0:w]
    N = open[0][0]
    if dist_est(N,goal,1,scale) > dist_est(bestEver,goal,1,scale):
      flag = False
    else:
      bestEver = N

beamSearch('S','G',2,5)
```
